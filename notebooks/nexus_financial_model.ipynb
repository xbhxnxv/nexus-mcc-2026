## 4. Operating position and the reconciliation issue

Revenue minus OPEX for each scenario, using the revised £4.5m OPEX.

This is where the source drafts don't line up. The Executive Summary states a £500k annual gap covered by the £3.2m reserve for five years. That only holds if revenue is £4.0m. The auditable bottom-up base case reaches £2.865m, which leaves a ~£1.6m gap and exhausts the reserve inside two years. The model uses the defensible bottom-up figure and flags the gap rather than smoothing it over.

```python
roi_scenarios('revised', a)
```

```python
cash_runway('revised', 'base', a)
```
